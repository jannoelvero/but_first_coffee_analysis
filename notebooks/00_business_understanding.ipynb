## 6. Competitive Price Index

For comparable products:

\[
Price\ Index = \frac{BFC\ Comparable\ Price}{Competitor\ Comparable\ Price} \times 100
\]

Interpretation:

- **Below 100** → BFC is priced below the competitor
- **100** → approximately equal pricing
- **Above 100** → BFC is priced above the competitor

Price comparisons will only be made between reasonably comparable products, categories, sizes, and formats.

Regular prices and promotional prices will be analyzed separately.

## 7. Statistical Hypotheses

### H1 — Matched Competitive Price Differences

**Null Hypothesis (H₀):**  
There is no systematic difference in representative price ranks among But First, Coffee, Starbucks Philippines, and The Coffee Bean & Tea Leaf Philippines across directly matched product families.

**Alternative Hypothesis (H₁):**  
There is a systematic difference in representative price ranks among the three brands across directly matched product families.

**Planned test:**
- Friedman test for matched product-family prices
- Post-hoc pairwise Wilcoxon signed-rank tests where justified, with multiple-comparison adjustment

**Methodological note:**  
Whole-menu price distributions will be used descriptively rather than as the primary inferential comparison because differences in product and category composition may confound cross-brand price comparisons.

---

### H2 — Individual Customer Rating Differences

**Null Hypothesis (H₀):**  
There is no statistically significant difference in individual customer-rating distributions among BFC, Starbucks Philippines, and CBTL Philippines.

**Alternative Hypothesis (H₁):**  
At least one brand has a statistically different individual customer-rating distribution.

Potential test:
- Kruskal-Wallis test

**Important:** This hypothesis will only be tested if adequate individual review-level rating data are available for all three brands.

## 10. Analytical Limitations

### Profitability
Actual profitability cannot be calculated because internal COGS, product costs, labor costs, transaction volumes, discount funding, delivery commissions, and contribution margins are unavailable.

### Revenue
Public data do not provide complete branch-level transaction counts or sales revenue. Revenue opportunities therefore represent analytical opportunities rather than forecasts of actual incremental revenue.

### Customer Reviews
Online reviews may contain self-selection bias and may not represent the entire customer population.

### Platform Differences
Customer behavior and ratings may differ between delivery platforms, Google-origin reviews, and in-store experiences.

### Aggregate vs Individual Ratings
Branch-level aggregate ratings must not be treated as individual customer ratings.

### Menu Comparability
Products may differ in size, ingredients, preparation, positioning, and format. Price comparisons will therefore use comparable products or clearly defined categories whenever possible.

### Store Counts
Public store-count information may differ according to date, source, store format, and definition.

### External Analysis
All strategic recommendations generated from this project should ultimately be validated against BFC's internal POS, product-cost, transaction, customer, and branch-performance data.
